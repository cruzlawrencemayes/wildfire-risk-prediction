```python
from pathlib import Path
import pandas as pd

DATA_FILENAME = "NM_Wildfires.csv"  # only the file name
FALLBACK_DIR = Path(r"C:\Users\clawr\OneDrive\Desktop\Dmining Phase 4\Datasets\Datasets")

cwd = Path(".").resolve()
fpath_cwd = cwd / DATA_FILENAME
fpath_fb = FALLBACK_DIR / DATA_FILENAME

if fpath_cwd.exists():
    fpath = fpath_cwd; source = "CWD"
elif fpath_fb.exists():
    fpath = fpath_fb; source = "FALLBACK_DIR"
else:
    raise FileNotFoundError(f"Could not find {DATA_FILENAME} in {cwd} or {FALLBACK_DIR}")

print("Using file from:", source)
print("Path:", fpath)

df = pd.read_csv(fpath, low_memory=False)
print("Shape:", df.shape)
display(df.head(3))
```

1. Define the target.
We predict whether a fire is “large” (≥ 100 acres). This is operationally meaningful and available for most rows.
To avoid leakage, we will later drop FIRE_SIZE from features after creating the label. This cell creates LargeFire and prints class balance

```python
import numpy as np
import pandas as pd

assert 'FIRE_SIZE' in df.columns, "FIRE_SIZE not found in df."

THRESH = 100.0  # consider sensitivity runs at 300 or 1000 in your analysis section
df = df.copy()
df['LargeFire'] = (df['FIRE_SIZE'].fillna(0) >= THRESH).astype(int)

print("Rows:", len(df))
print("LargeFire counts:\n", df['LargeFire'].value_counts(dropna=False))
print("Positive rate (LargeFire=1):", df['LargeFire'].mean())
```

--Notes--
Class imbalance is ~6.3% positives, so we’ll keep class_weight='balanced' in models.

2. Select features (no leakage).
We drop FIRE_SIZE (source of the label) to avoid leakage.
We keep a compact set supported by our EDA:

Numeric: FIRE_YEAR, DISCOVERY_DOY, LATITUDE, LONGITUDE, ELEVATION (if present)

Categorical: STAT_CAUSE_DESCR (cause), plus Season derived from DISCOVERY_DOY
We’ll impute/encode later in a pipeline.

```python
import pandas as pd
import numpy as np

df = df.copy()

# Optional derived feature: Season from day-of-year (non-leaky)
if "DISCOVERY_DOY" in df.columns:
    bins = [-1, 59, 151, 243, 334, 366]  # winter/spring/summer/fall/winter wrap
    labels = ["Winter","Spring","Summer","Fall","Winter2"]
    df["Season"] = pd.cut(df["DISCOVERY_DOY"], bins=bins, labels=labels, include_lowest=True)

# Define candidate features and keep only those that actually exist
num_candidates = ["FIRE_YEAR","DISCOVERY_DOY","LATITUDE","LONGITUDE","ELEVATION"]
cat_candidates = ["STAT_CAUSE_DESCR","Season"]

num_features_present = [c for c in num_candidates if c in df.columns]
cat_features_present = [c for c in cat_candidates if c in df.columns]

# Build X and y (drop leakage columns and the target)
leakage_cols = ["FIRE_SIZE"]  # direct source of target
X = df.drop(columns=[c for c in leakage_cols if c in df.columns] + ["LargeFire"])
y = df["LargeFire"]

# Keep only our selected feature set
keep = num_features_present + cat_features_present
X = X[keep]

print("Selected features:", keep)
print("X shape:", X.shape, "| y shape:", y.shape)
print("\nTop nulls in X:")
print(X.isna().sum().sort_values(ascending=False).head(10))
```

3. Split & preprocessing.
We do a stratified train/test split (preserves the 6.3% positive ratio).
Preprocessing uses a ColumnTransformer so the pipeline is reproducible:

Numeric → (no impute needed here) + StandardScaler (helps LR/KNN).

Categorical → OneHotEncoder(handle_unknown='ignore').
Keeping preprocessing inside the pipeline prevents data leakage.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Identify numeric vs categorical from your selected features
num_features = ["FIRE_YEAR","DISCOVERY_DOY","LATITUDE","LONGITUDE"]
cat_features = [c for c in ["STAT_CAUSE_DESCR","Season"] if c in X.columns]

# Stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# Preprocessing
numeric_pipe = Pipeline(steps=[
    ("scale", StandardScaler())
])

categorical_pipe = Pipeline(steps=[
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocess = ColumnTransformer(
    transformers=[
        ("num", numeric_pipe, num_features),
        ("cat", categorical_pipe, cat_features)
    ]
)

print("Train shapes:", X_train.shape, y_train.shape, "| Test shapes:", X_test.shape, y_test.shape)
print("Num features:", num_features)
print("Cat features:", cat_features)
```

4. Logistic Regression (baseline). - Let's train our first classifier
Why: fast, strong linear baseline, interpretable.
How: Wrap preprocessing + model in a single Pipeline to avoid leakage.
Eval: 5-fold stratified CV on the train set (F1), then test metrics (Accuracy, Precision, Recall, F1) and a confusion matrix. We use class_weight='balanced' due to the ~6.3% positive rate.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt

logreg_pipe = Pipeline(steps=[
    ("prep", preprocess),
    ("clf", LogisticRegression(max_iter=300, class_weight="balanced"))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(logreg_pipe, X_train, y_train, cv=cv, scoring="f1")
print("LogReg CV F1:", cv_scores.mean(), "+/-", cv_scores.std())

logreg_pipe.fit(X_train, y_train)
y_pred = logreg_pipe.predict(X_test)

acc = accuracy_score(y_test, y_pred)
pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred, average="binary", zero_division=0)
print(f"Test  Acc: {acc:.3f} | Prec: {pr:.3f} | Rec: {rc:.3f} | F1: {f1:.3f}")
print("\nClassification report:\n", classification_report(y_test, y_pred, digits=3))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.title("Confusion Matrix — Logistic Regression")
plt.tight_layout(); plt.show()
```

--Notes--
These results make sense for an imbalanced label with class_weight='balanced':

High recall (0.686) on the positive class but low precision (0.148) → the model is aggressive about flagging large fires (many false positives).

We’ll see if tree-based models can improve precision/F1.

5. Random Forest.
Why: non-linear, handles mixed features well, often improves precision vs. linear models.
How: Same Pipeline pattern (preprocessing + model).
Eval: 5-fold CV (F1) on train, then test metrics and confusion matrix.
--Notes--
We use class_weight='balanced_subsample' to address the 6.3% positive rate.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt

rf_pipe = Pipeline(steps=[
    ("prep", preprocess),
    ("clf", RandomForestClassifier(
        n_estimators=400,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced_subsample"
    ))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(rf_pipe, X_train, y_train, cv=cv, scoring="f1")
print("RF CV F1:", cv_scores.mean(), "+/-", cv_scores.std())

rf_pipe.fit(X_train, y_train)
y_pred = rf_pipe.predict(X_test)

acc = accuracy_score(y_test, y_pred)
pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred, average="binary", zero_division=0)
print(f"Test  Acc: {acc:.3f} | Prec: {pr:.3f} | Rec: {rc:.3f} | F1: {f1:.3f}")
print("\nClassification report:\n", classification_report(y_test, y_pred, digits=3))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.title("Confusion Matrix — Random Forest")
plt.tight_layout(); plt.show()
```

--Note--
This has great contrast to Logistic Regression:

RF: much higher precision (0.552) but low recall (0.142) on the positive class → it’s conservative (few false positives, many false negatives).

LR: lower precision (0.148) but higher recall (0.686) → aggressive (flags many positives, more false alarms).

6. KNN (distance-based baseline)
Why: simple distance-based baseline; good contrast with LR and RF.
How: We keep the same preprocessing (scaling helps KNN).
Eval: 5-fold CV (F1), then test metrics + confusion matrix.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt

knn_pipe = Pipeline(steps=[
    ("prep", preprocess),                 # includes StandardScaler + OneHot
    ("clf", KNeighborsClassifier(n_neighbors=15))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(knn_pipe, X_train, y_train, cv=cv, scoring="f1")
print("KNN CV F1:", cv_scores.mean(), "+/-", cv_scores.std())

knn_pipe.fit(X_train, y_train)
y_pred = knn_pipe.predict(X_test)

acc = accuracy_score(y_test, y_pred)
pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred, average="binary", zero_division=0)
print(f"Test  Acc: {acc:.3f} | Prec: {pr:.3f} | Rec: {rc:.3f} | F1: {f1:.3f}")
print("\nClassification report:\n", classification_report(y_test, y_pred, digits=3))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.title("Confusion Matrix — KNN (k=15)")
plt.tight_layout(); plt.show()
```

7. Model comparison table
This cell evaluates the three trained pipelines (LogReg, RF, KNN) consistently and returns a table with:

CV F1 (mean ± std) on the training split (5-fold stratified)

Test Accuracy, Precision, Recall, and F1
We’ll use this table directly in the Feature Summary section.

```python
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def eval_model(name, pipe, X_train, y_train, X_test, y_test, cv):
    cv_scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="f1")
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred, average="binary", zero_division=0)
    return {
        "Model": name,
        "CV_F1_mean": cv_scores.mean(),
        "CV_F1_std": cv_scores.std(),
        "Test_Accuracy": acc,
        "Test_Precision": pr,
        "Test_Recall": rc,
        "Test_F1": f1
    }

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = []
results.append(eval_model("LogisticRegression", logreg_pipe, X_train, y_train, X_test, y_test, cv))
results.append(eval_model("RandomForest", rf_pipe, X_train, y_train, X_test, y_test, cv))
results.append(eval_model("KNN (k=15)", knn_pipe, X_train, y_train, X_test, y_test, cv))

cmp_df = pd.DataFrame(results).sort_values("Test_F1", ascending=False)
cmp_df
```

8. Classifier comparison 

We trained three models on the NM Wildfires dataset using a consistent preprocessing pipeline (scaling numeric features and one-hot encoding categorical features). Logistic Regression achieved the best test F1 (≈0.244) with high recall (≈0.686) but low precision (≈0.148), making it effective at catching large fires but with many false positives. Random Forest delivered much higher precision (≈0.552) but low recall (≈0.142) and similar F1 (≈0.226), favoring fewer false alarms at the cost of missing many positives. KNN (k=15) underperformed (F1 ≈0.110) given the class imbalance and feature geometry. These outcomes show a clear precision–recall trade-off: LR is better if recall is critical (e.g., early detection), while RF is preferable when false alarms are costly (e.g., limited resources). We recommend Logistic Regression with calibrated thresholding to balance recall and precision for the team’s chosen objective.

9. Tuning-- Threshold tuning for LR.
    
The default threshold of 0.5 isn’t optimal for imbalanced data. This cell sweeps thresholds and prints the best F1 and a few checkpoints so you can pick a setting that matches your team’s goal (e.g., recall ≥ 0.50 with better precision).

```python
import numpy as np
from sklearn.metrics import precision_recall_fscore_support

# get probabilities from the already-fitted LR pipeline
y_prob = logreg_pipe.predict_proba(X_test)[:, 1]

def metrics_at(th):
    y_hat = (y_prob >= th).astype(int)
    pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_hat, average="binary", zero_division=0)
    return pr, rc, f1

# sweep thresholds
ths = np.linspace(0.05, 0.95, 19)
rows = []
for th in ths:
    pr, rc, f1 = metrics_at(th)
    rows.append((th, pr, rc, f1))
tune = pd.DataFrame(rows, columns=["threshold","precision","recall","f1"]).sort_values("threshold")
display(tune.round(3).head(10))
display(tune.round(3).tail(10))

best = tune.loc[tune["f1"].idxmax()]
print("\nBest F1 row:\n", best.round(3))

# Example: pick the smallest threshold that gets recall >= 0.50 with better precision than default
candidates = tune[tune["recall"] >= 0.50].sort_values(["precision","f1"], ascending=[False,False]).head(5)
print("\nTop candidates with recall >= 0.50:\n", candidates.round(3))
```

10. Apply tuned threshold (0.55)
We now apply the tuned decision threshold of 0.55 to generate new predictions and plot an updated confusion matrix.
This version of Logistic Regression offers a better trade-off between precision and recall than the default 0.5 cutoff.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, precision_recall_fscore_support

THRESH_TUNED = 0.55  # from Step 9

y_prob = logreg_pipe.predict_proba(X_test)[:, 1]
y_pred_tuned = (y_prob >= THRESH_TUNED).astype(int)

pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred_tuned, average="binary", zero_division=0)
print(f"Tuned threshold = {THRESH_TUNED}")
print(f"Precision: {pr:.3f} | Recall: {rc:.3f} | F1: {f1:.3f}")

cm = confusion_matrix(y_test, y_pred_tuned)
ConfusionMatrixDisplay(cm).plot(cmap="viridis")
plt.title(f"Confusion Matrix — LR (threshold={THRESH_TUNED})")
plt.tight_layout(); plt.show()
```
